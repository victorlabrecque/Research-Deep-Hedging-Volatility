Simulation of the joint spot and option surface:

- Option 1: historica resampling: Filtered Historical Simulation (FHS) extended to a joint state. Resample actual paired shocks (same as the paper on dynamic portfolio management with options):
    - Spot leg (GARCH): get standardized residuals
    - Latent leg: get latent state time serie from historical options
    - Pair them by date: each date has their joint shock
    - Simulate by resampling
    - Drawdown: resampling each day separately destroys volatility clustering and short term dependance. Instead resample blocks (e.g. 5 or 10 days blocks)
    - Keep a parametric mean reversion, to avoid random walk
- Option 2: Joint parametric distribution (Gaussian copula, fixed correlation matrix)
    - Spot leg: GARCH standardized residuals, same as option 1
    - Latent leg: parametric mean-reverting process of latent state series
    - Advantages: fewer parameter to estimate, can simulate shock outside historical range: useful for stress testing
    - Drawback: assume fixed symmetric dependence structure: miss asymetry, tails co-dependence (leverage effect)
- Option 3: Mix of both, simulate option surfaces from a weight between Gaussian innovation and historical resampling, then simulate the spot.
